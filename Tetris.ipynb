```python
grid = [ [ " " for i in range(16) ] for i in range(10) ]
```

```python
for n, column in enumerate(grid):
    print(n, column)
```

```python
drop_column = 8

try:
    n = grid[drop_column].index("*")
    n = n - 1
except:
    n = -1

grid[drop_column][n] = "*"

for n, column in enumerate(grid):
    print(n, column)
```

```python
def drop_block(column_number):
    try:
        n = grid[column_number].index("*")
        n = n - 1
    except:
        n = -1
    grid[column_number][n] = "*"
```

```python
drop_block(5)
```

```python
drop_block(7)
```

```python
for n, column in enumerate(grid):
    print(n, column)
```
